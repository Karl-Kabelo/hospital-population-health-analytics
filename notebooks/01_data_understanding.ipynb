# Notebook 01: Synthea Data Acquisition, Schema Exploration & Entity Relationships
**Project:** Hospital Population Health & Clinical Operations Analytics Platform  
**Dataset Provenance:** Synthetic patient records modeled on [Synthea](https://github.com/synthetichealth/synthea) ([CSV File Data Dictionary](https://github.com/synthetichealth/synthea/wiki/CSV-File-Data-Dictionary)).  
*Disclaimer: All records in this notebook are 100% synthetic and do not represent real patients or actual hospital statistics.*


```python
import os
import pandas as pd

raw_dir = '../data/raw'
files = ['patients.csv', 'encounters.csv', 'conditions.csv', 'observations.csv',
         'medications.csv', 'procedures.csv', 'organizations.csv', 'providers.csv']
summary = []
for f in files:
    df = pd.read_csv(os.path.join(raw_dir, f), dtype=str)
    summary.append({'file': f, 'rows': len(df), 'columns': len(df.columns), 'sample_cols': ', '.join(df.columns[:5])})
print(pd.DataFrame(summary).to_string(index=False))
```

## Primary Key & Foreign Key Relationship Audit
- `patients.csv` (`Id` PK) -> linked by `PATIENT` across `encounters`, `conditions`, `observations`, `medications`, and `procedures`.
- `encounters.csv` (`Id` PK) -> linked by `ENCOUNTER` across clinical event tables (`NULL` on 270 annual Quality-of-Life survey rows in `observations.csv`).
