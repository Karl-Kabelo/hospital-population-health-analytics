# Notebook 03: Population Health & Clinical Operations Analysis
**Objective:** Query `hospital_analytics.db` to evaluate hospital utilization, length of stay, 30-day inpatient readmissions, and high-risk cohorts.


```python
import sqlite3
import pandas as pd
conn = sqlite3.connect('../data/processed/hospital_analytics.db')
kpi_df = pd.read_sql_query('''
    SELECT COUNT(DISTINCT patient_id) AS unique_patients,
           COUNT(encounter_id) AS total_encounters,
           ROUND(COUNT(encounter_id)*1.0/COUNT(DISTINCT patient_id), 2) AS avg_enc_per_pat,
           ROUND(AVG(los_hours), 2) AS avg_los_hours,
           ROUND(SUM(is_emergency)*100.0/COUNT(*), 2) AS emergency_rate_pct
    FROM fact_encounter
''', conn)
print(kpi_df.to_string(index=False))
```
